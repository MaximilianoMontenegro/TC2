## Introducción teórica

La **síntesis de redes** consiste en determinar la topología de un circuito eléctrico y los valores de sus componentes a partir de una función de impedancia $Z(s)$ o admitancia $Y(s)$, relacionadas mediante:

$$
Z(s)=\frac{1}{Y(s)}
$$

### Método de Foster

Se basa en la **descomposición en fracciones parciales** de la función de inmitancia, permitiendo identificar los elementos que componen la red.

Para una red LC, una expresión general es:

$$
F(s)=K_\infty s+\frac{K_0}{s}+
\sum_{i=1}^{n}\frac{K_i s}{s^2+\omega_i^2}
$$

La síntesis puede realizarse en **serie**, trabajando con $Z(s)$, o en **derivación**, trabajando con $Y(s)$.

### Método de Cauer

Se basa en la **remoción sucesiva de elementos**, mediante el desarrollo en fracciones continuas. Las remociones pueden realizarse en $s=0$ (DC) o en $s\to\infty$.

Para una impedancia, los coeficientes de remoción LC se calculan mediante:

**Remoción en infinito (inductor serie):**

$$
L=\lim_{s\to\infty}\frac{Z(s)}{s}
$$

**Remoción en cero (capacitor serie):**

$$
\frac{1}{C}=\lim_{s\to0}sZ(s)
$$

Una vez removido el elemento, se obtiene la función residual:

$$
Z_1(s)=Z(s)-Z_{\mathrm{removido}}(s)
$$

Luego se invierte para continuar la síntesis:

$$
Y_1(s)=\frac{1}{Z_1(s)}
$$

Ambos métodos permiten obtener diferentes topologías circuitales que satisfacen una misma función de inmitancia.



## 1.a) Síntesis de Foster en serie

Se parte de la función de inmitancia:

$$
F(s)=\frac{s^4+4s^2+3}{s(s^2+2)}
$$

Para realizar la síntesis mediante el método de **Foster en serie**, se utiliza la descomposición en fracciones parciales:

$$
F(s)=\frac{K_0}{s}+K_\infty s+
\sum_{i=1}^{n}\frac{2K_i s}{s^2+\omega_i^2}
$$

### Cálculo de los coeficientes

**1. Remoción del polo en el origen**

Se determina el coeficiente $K_0$:

$$
K_0=\lim_{s\to0}sF(s)
$$

Sustituyendo:

$$
K_0=\lim_{s\to0}
s\frac{s^4+4s^2+3}{s(s^2+2)}
$$

$$
\boxed{K_0=\frac32}
$$

**2. Remoción del polo en infinito**

Se calcula el coeficiente $K_\infty$:

$$
K_\infty=\lim_{s\to\infty}\frac{F(s)}{s}
$$

$$
K_\infty=\lim_{s\to\infty}
\frac{s^4+4s^2+3}{s^2(s^2+2)}
$$

$$
\boxed{K_\infty=1}
$$

**3. Cálculo del residuo del polo en $s^2=-2$**

Se determina el coeficiente $2K_1$:

$$
2K_1=\lim_{s^2\to-2}
\frac{s^2+2}{s}F(s)
$$

Sustituyendo:

$$
2K_1=\lim_{s^2\to-2}
\frac{s^4+4s^2+3}{s^2}
$$

$$
2K_1=\frac{4-8+3}{-2}
$$

$$
\boxed{2K_1=\frac12}
$$

### Descomposición final

Reemplazando los coeficientes obtenidos:

$$
\boxed{
F(s)=\frac{3}{2s}+s+\frac{\frac12s}{s^2+2}
}
$$

### Determinación de los componentes

Como se trabaja con la impedancia $Z(s)$, cada término representa una impedancia conectada en serie.

**Capacitor serie $C_1$:**

$$
Z_{C_1}(s)=\frac{1}{sC_1}=\frac{3}{2s}
$$

Por lo tanto:

$$
\boxed{C_1=\frac23\ \mathrm{F}}
$$

**Inductor serie $L_1$:**

$$
Z_{L_1}(s)=sL_1=s
$$

Por lo tanto:

$$
\boxed{L_1=1\ \mathrm{H}}
$$

**Circuito LC en paralelo:**

$$
Z_{LC}(s)=\frac{\frac12s}{s^2+2}
$$

La impedancia de un circuito LC en paralelo es:

$$
Z_{LC}(s)=
\frac{\frac{s}{C_2}}{s^2+\frac{1}{L_2C_2}}
$$

Comparando numeradores:

$$
\frac{1}{C_2}=\frac12
$$

$$
\boxed{C_2=2\ \mathrm{F}}
$$

Comparando denominadores:

$$
\frac{1}{L_2C_2}=2
$$

$$
L_2=\frac{1}{2C_2}
$$

$$
\boxed{L_2=\frac14\ \mathrm{H}}
$$

### Resultados

| Componente | Valor | Configuración |
|:---:|:---:|:---|
| $C_1$ | $2/3\ \mathrm{F}$ | Serie |
| $L_1$ | $1\ \mathrm{H}$ | Serie |
| $C_2$ | $2\ \mathrm{F}$ | Paralelo con $L_2$ |
| $L_2$ | $1/4\ \mathrm{H}$ | Paralelo con $C_2$ |

La topología resultante consiste en un capacitor $C_1$ y un inductor $L_1$ conectados en serie, seguidos de una rama resonante formada por $L_2$ y $C_2$ en paralelo.

## 1.c) Síntesis de Cauer por remoción en infinito

Se parte de la función de impedancia:

$$
Z(s)=\frac{s^4+4s^2+3}{s(s^2+2)}
$$

El método de Cauer consiste en realizar remociones sucesivas de elementos, alternando entre impedancia y admitancia.

En este caso, se comienza con una **remoción en infinito**.

### 1. Primera remoción: inductor serie

Se calcula el coeficiente correspondiente al polo en infinito:

$$
K_\infty=\lim_{s\to\infty}\frac{Z(s)}{s}
$$

$$
K_\infty=
\lim_{s\to\infty}
\frac{s^4+4s^2+3}{s^2(s^2+2)}
$$

$$
\boxed{K_\infty=1}
$$

Por lo tanto, se remueve un inductor serie:

$$
\boxed{L_1=1\ \mathrm{H}}
$$

La impedancia residual es:

$$
Z_1(s)=Z(s)-sL_1
$$

$$
Z_1(s)=
\frac{s^4+4s^2+3}{s^3+2s}-s
$$

$$
Z_1(s)=
\frac{s^4+4s^2+3-s^4-2s^2}{s^3+2s}
$$

$$
\boxed{
Z_1(s)=\frac{2s^2+3}{s(s^2+2)}
}
$$

### 2. Segunda remoción: capacitor en derivación

Se invierte la impedancia residual:

$$
Y_1(s)=\frac{1}{Z_1(s)}
$$

$$
Y_1(s)=\frac{s(s^2+2)}{2s^2+3}
$$

Se calcula el coeficiente en infinito:

$$
C_1=\lim_{s\to\infty}\frac{Y_1(s)}{s}
$$

$$
C_1=\lim_{s\to\infty}
\frac{s^2+2}{2s^2+3}
$$

$$
\boxed{C_1=\frac12\ \mathrm{F}}
$$

Se remueve el capacitor:

$$
Y_2(s)=Y_1(s)-\frac{s}{2}
$$

$$
Y_2(s)=
\frac{s(s^2+2)}{2s^2+3}-\frac{s}{2}
$$

$$
Y_2(s)=
\frac{2s^3+4s-2s^3-3s}{2(2s^2+3)}
$$

$$
\boxed{
Y_2(s)=\frac{s}{4s^2+6}
}
$$

### 3. Tercera remoción: inductor serie

Se invierte nuevamente:

$$
Z_2(s)=\frac{1}{Y_2(s)}
$$

$$
Z_2(s)=\frac{4s^2+6}{s}
$$

Separando términos:

$$
Z_2(s)=4s+\frac6s
$$

Se calcula el coeficiente en infinito:

$$
L_2=\lim_{s\to\infty}\frac{Z_2(s)}{s}
$$

$$
\boxed{L_2=4\ \mathrm{H}}
$$

Se remueve el inductor:

$$
Z_3(s)=Z_2(s)-4s
$$

$$
\boxed{Z_3(s)=\frac6s}
$$

### 4. Cuarta remoción: capacitor en derivación

Se invierte la impedancia residual:

$$
Y_3(s)=\frac1{Z_3(s)}
$$

$$
Y_3(s)=\frac{s}{6}
$$

Comparando con la admitancia de un capacitor:

$$
Y_C(s)=sC
$$

Se obtiene:

$$
\boxed{C_2=\frac16\ \mathrm{F}}
$$

Con esta última remoción se completa la síntesis.

### Resultados

| Componente | Valor | Conexión |
|:---:|:---:|:---|
| $L_1$ | $1\ \mathrm{H}$ | Serie |
| $C_1$ | $1/2\ \mathrm{F}$ | Derivación |
| $L_2$ | $4\ \mathrm{H}$ | Serie |
| $C_2$ | $1/6\ \mathrm{F}$ | Derivación |

La red resultante presenta una **estructura de escalera LC**, comenzando con un inductor serie, seguido de un capacitor en derivación, otro inductor serie y un capacitor final en derivación.



## Ejercicio 3 — Síntesis de una red RLC

Se propone la siguiente función de impedancia:

$$
Z(s)=\frac{s^2+6s+8}{s^2+6s+5}
$$

La red está formada por una resistencia $R_1$ en serie con una rama $R_2\parallel L$ y otra rama $R_3\parallel C$.

Por lo tanto:

$$
Z(s)=R_1+Z_{RL}(s)+Z_{RC}(s)
$$

### 1. Factorización de la impedancia

Factorizando numerador y denominador:

$$
Z(s)=\frac{(s+2)(s+4)}{(s+1)(s+5)}
$$

### 2. Análisis en infinito y en cero

Se calcula el valor de la impedancia en infinito:

$$
K_\infty=\lim_{s\to\infty}Z(s)=1
$$

En alta frecuencia, el inductor se comporta como circuito abierto y el capacitor como cortocircuito.

Por lo tanto:

$$
\boxed{R_1+R_2=1}
$$

Se calcula la impedancia en cero:

$$
K_0=\lim_{s\to0}Z(s)=\frac85
$$

En corriente continua, el inductor se comporta como cortocircuito y el capacitor como circuito abierto.

Por lo tanto:

$$
\boxed{R_1+R_3=\frac85}
$$

### 3. Cálculo de los polos y residuos

Los polos de la función son:

$$
s_1=-1,\qquad s_2=-5
$$

El residuo correspondiente a $s=-1$ es:

$$
A_1=\lim_{s\to-1}(s+1)Z(s)
$$

$$
A_1=\frac{(-1+2)(-1+4)}{-1+5}
$$

$$
\boxed{A_1=\frac34}
$$

El residuo correspondiente a $s=-5$ es:

$$
A_2=\lim_{s\to-5}(s+5)Z(s)
$$

$$
A_2=\frac{(-5+2)(-5+4)}{-5+1}
$$

$$
\boxed{A_2=-\frac34}
$$

Realizando la descomposición en fracciones simples:

$$
\boxed{
Z(s)=1+\frac{3}{4(s+1)}-\frac{3}{4(s+5)}
}
$$

### 4. Determinación de la rama RC

La impedancia de una resistencia en paralelo con un capacitor es:

$$
Z_{RC}(s)=\frac{R_3}{1+sR_3C}
$$

Reordenando:

$$
Z_{RC}(s)=\frac{1/C}{s+1/(R_3C)}
$$

Comparando con el término de residuo positivo:

$$
\frac{3/4}{s+1}
=
\frac{1/C}{s+1/(R_3C)}
$$

Igualando numeradores:

$$
\frac1C=\frac34
$$

$$
\boxed{C=\frac43\ \mathrm{F}}
$$

Igualando los términos constantes del denominador:

$$
\frac1{R_3C}=1
$$

$$
R_3=\frac1C
$$

$$
\boxed{R_3=\frac34\ \Omega}
$$

### 5. Determinación de R1 y R2

Utilizando la condición en cero:

$$
R_1+R_3=\frac85
$$

Reemplazando $R_3=3/4$:

$$
R_1=\frac85-\frac34
$$

$$
R_1=\frac{32-15}{20}
$$

$$
\boxed{R_1=\frac{17}{20}\ \Omega}
$$

Utilizando la condición en infinito:

$$
R_1+R_2=1
$$

$$
R_2=1-\frac{17}{20}
$$

$$
\boxed{R_2=\frac3{20}\ \Omega}
$$

### 6. Determinación de la inductancia

La impedancia de la rama RL es:

$$
Z_{RL}(s)=\frac{sLR_2}{sL+R_2}
$$

Dividiendo numerador y denominador por $L$:

$$
Z_{RL}(s)=\frac{sR_2}{s+R_2/L}
$$

El polo de esta rama corresponde a:

$$
s=-5
$$

Por lo tanto:

$$
\frac{R_2}{L}=5
$$

Despejando:

$$
L=\frac{R_2}{5}
$$

Con $R_2=3/20$:

$$
\boxed{L=\frac3{100}\ \mathrm{H}}
$$

### 7. Verificación de los valores

Es necesario comprobar que la suma de las impedancias de las ramas reproduce la función original.

La rama RC es:

$$
Z_{RC}(s)=\frac{3/4}{s+1}
$$

La rama RL, con los valores anteriores, es:

$$
Z_{RL}(s)=\frac{3s/20}{s+5}
$$

La impedancia total obtenida sería:

$$
Z_{\mathrm{calc}}(s)=
\frac{17}{20}+
\frac{3s}{20(s+5)}+
\frac{3}{4(s+1)}
$$

Esta expresión **no coincide** con la impedancia propuesta.

Por lo tanto, los valores calculados mediante la asociación directa del residuo positivo a la rama RC no constituyen una solución válida. Es necesario realizar una descomposición diferente que tenga en cuenta la contribución constante de la rama RL.

### 8. Descomposición correcta

Se utiliza la identidad:

$$
-\frac{3}{4(s+5)}
=
\frac{3s}{20(s+5)}-\frac3{20}
$$

Sin embargo, al combinar esta expresión con el término constante se comprueba que no se obtiene una realización compatible con los valores anteriores.

Para determinar una realización válida, se parte de:

$$
Z(s)=R_1+\frac{R_2s}{s+a}+\frac{b}{s+c}
$$

donde:

$$
a=\frac{R_2}{L},\qquad
b=\frac1C,\qquad
c=\frac1{R_3C}
$$

Se asignan los polos:

$$
a=5,\qquad c=1
$$

De la descomposición:

$$
Z(s)=1+\frac{3}{4(s+1)}-\frac{3}{4(s+5)}
$$

Para la rama RL:

$$
\frac{R_2s}{s+5}
=
R_2-\frac{5R_2}{s+5}
$$

Igualando residuos:

$$
5R_2=\frac34
$$

$$
R_2=\frac3{20}
$$

Por la condición en infinito:

$$
R_1=1-R_2=\frac{17}{20}
$$

La rama RC conserva:

$$
C=\frac43,\qquad R_3=\frac34
$$

Y:

$$
L=\frac{R_2}{5}=\frac3{100}
$$

La verificación correcta es:

$$
Z(s)=\frac{17}{20}
+\frac{3s}{20(s+5)}
+\frac{3}{4(s+1)}
$$

$$
\boxed{
Z(s)=\frac{s^2+6s+8}{s^2+6s+5}
}
$$

### 9. Valores finales

| Componente | Valor |
|---|---|
| $R_1$ | $17/20\ \Omega$ |
| $R_2$ | $3/20\ \Omega$ |
| $R_3$ | $3/4\ \Omega$ |
| $L$ | $3/100\ \mathrm{H}$ |
| $C$ | $4/3\ \mathrm{F}$ |

La impedancia equivalente de la red coincide con la función propuesta.
