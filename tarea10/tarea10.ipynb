### Datos del problema

Se desea sintetizar un filtro pasabajos Chebyshev de tipo I, de tercer orden, con las siguientes características:

- Orden del filtro: $n=3$
- Rizado máximo en la banda de paso: $\alpha_p=1\,\mathrm{dB}$
- Frecuencia angular normalizada: $\omega_p=1\,\mathrm{rad/s}$
- Impedancia de referencia: $Z_0=1\,\Omega$
- Resistencia de carga: $R_L=1\,\Omega$



## Parámetros de dispersión (parámetros S)

Los parámetros de dispersión permiten caracterizar el comportamiento de un cuadripolo mediante las ondas incidentes y reflejadas en sus puertos.

En un cuadripolo de dos puertos se definen:

- $a_1$: onda incidente en el puerto 1.
- $b_1$: onda reflejada o saliente del puerto 1.
- $a_2$: onda incidente en el puerto 2.
- $b_2$: onda reflejada o saliente del puerto 2.

La relación matricial entre las ondas es:

$$
\begin{bmatrix}
b_1\\
b_2
\end{bmatrix}
=
\begin{bmatrix}
S_{11} & S_{12}\\
S_{21} & S_{22}
\end{bmatrix}
\begin{bmatrix}
a_1\\
a_2
\end{bmatrix}
$$

Desarrollando:

$$
b_1=S_{11}a_1+S_{12}a_2
$$

$$
b_2=S_{21}a_1+S_{22}a_2
$$

### Parámetro $S_{11}$ — Reflexión en la entrada

Se define como:

$$
\boxed{S_{11}=\left.\frac{b_1}{a_1}\right|_{a_2=0}}
$$


El parámetro $S_{11}$ indica qué proporción de la onda incidente se refleja en la entrada del cuadripolo.

Para una impedancia de referencia real $Z_0$:

$$
\boxed{S_{11}=\frac{Z_{\mathrm{in}}-Z_0}{Z_{\mathrm{in}}+Z_0}}
$$

### Parámetro $S_{21}$ — Transmisión directa

Se define como:

$$
\boxed{S_{21}=\left.\frac{b_2}{a_1}\right|_{a_2=0}}
$$

Representa la transmisión desde el puerto 1 hacia el puerto 2, cuando el puerto 2 está adaptado.

### Parámetro $S_{12}$ — Transmisión inversa

Se define como:

$$
\boxed{S_{12}=\left.\frac{b_1}{a_2}\right|_{a_1=0}}
$$

Representa la transmisión desde el puerto 2 hacia el puerto 1, manteniendo adaptado el puerto 1.

Para un cuadripolo recíproco:

$$
\boxed{S_{12}=S_{21}}
$$

### Parámetro $S_{22}$ — Reflexión en la salida

Se define como:

$$
\boxed{S_{22}=\left.\frac{b_2}{a_2}\right|_{a_1=0}}
$$

El parámetro $S_{22}$ indica la reflexión que se produce cuando una onda incide desde el puerto 2.

Si $Z_{\mathrm{out}}$ es la impedancia vista desde el puerto 2, con el puerto 1 terminado en $Z_0$:

$$
\boxed{S_{22}=\frac{Z_{\mathrm{out}}-Z_0}{Z_{\mathrm{out}}+Z_0}}
$$

### Propiedades de los cuadripolos

**Reciprocidad**

En una red recíproca:

$$
S_{12}=S_{21}
$$

**Simetría**

En una red simétrica, con iguales impedancias de referencia:

$$
S_{11}=S_{22}
$$

**Conservación de potencia**

Para un cuadripolo ideal sin pérdidas, con impedancias de referencia reales:

$$
|S_{11}|^2+|S_{21}|^2=1
$$

Esto significa que la potencia incidente en el puerto 1 se distribuye entre la potencia reflejada y la transmitida.



# Resolución de los ejercicios

## Parte A — Síntesis de un filtro Chebyshev

### Punto 1 — Obtención de la impedancia de entrada

Se desea sintetizar un filtro pasabajos Chebyshev de tipo I, de tercer orden, con un rizado máximo de $1\,\mathrm{dB}$ en la banda de paso.

Los parámetros normalizados del filtro son:

- Orden: $n=3$
- Rizado máximo: $A_{\max}=1\,\mathrm{dB}$
- Frecuencia angular de borde de banda: $\omega_p=1\,\mathrm{rad/s}$
- Impedancia de referencia: $Z_0=1\,\Omega$
- Resistencia de carga: $R_L=1\,\Omega$

El objetivo es determinar la impedancia de entrada $Z_1(s)$ del cuadripolo, considerando que el puerto de salida está terminado en su impedancia de referencia.

### 2.1. Cálculo del factor de rizado

La respuesta de magnitud de un filtro Chebyshev de tipo I está dada por:

$$
|S_{21}(j\omega)|^2=
\frac{1}{1+\varepsilon^2 T_n^2(\omega/\omega_p)}
$$

donde:

- $\varepsilon$ es el factor de rizado.
- $T_n$ es el polinomio de Chebyshev de orden $n$.
- $\omega_p$ es la frecuencia angular del borde de la banda de paso.

El factor de rizado se calcula mediante:

$$
\varepsilon=\sqrt{10^{A_{\max}/10}-1}
$$

Reemplazando $A_{\max}=1\,\mathrm{dB}$:

$$
\varepsilon=\sqrt{10^{1/10}-1}
$$

Por lo tanto:

$$
\boxed{\varepsilon\approx0.50885}
$$

Este parámetro determina la amplitud del rizado de la respuesta Chebyshev dentro de la banda de paso.



### Polinomio de Chebyshev de tercer orden

Los polinomios de Chebyshev se pueden obtener mediante la siguiente relación de recurrencia:

$$
T_0(\omega)=1
$$

$$
T_1(\omega)=\omega
$$

$$
T_n(\omega)=2\omega T_{n-1}(\omega)-T_{n-2}(\omega)
$$

Para un filtro de tercer orden:

$$
T_2(\omega)=2\omega T_1(\omega)-T_0(\omega)
$$

$$
T_2(\omega)=2\omega^2-1
$$

Luego:

$$
T_3(\omega)=2\omega T_2(\omega)-T_1(\omega)
$$

$$
T_3(\omega)=2\omega(2\omega^2-1)-\omega
$$

Por lo tanto:

$$
\boxed{T_3(\omega)=4\omega^3-3\omega}
$$

### 2.3. Obtención de la respuesta en frecuencia

La magnitud al cuadrado del parámetro de transmisión está dada por:

$$
|S_{21}(j\omega)|^2=
\frac{1}{1+\varepsilon^2 T_3^2(\omega)}
$$

Sustituyendo el polinomio obtenido:

$$
|S_{21}(j\omega)|^2=
\frac{1}{1+\varepsilon^2(4\omega^3-3\omega)^2}
$$

Considerando que el cuadripolo es ideal y no presenta pérdidas:

$$
|S_{11}(j\omega)|^2+|S_{21}(j\omega)|^2=1
$$

Despejando:

$$
|S_{11}(j\omega)|^2=1-|S_{21}(j\omega)|^2
$$

Por lo tanto:

$$
\boxed{
|S_{11}(j\omega)|^2=
\frac{\varepsilon^2(4\omega^3-3\omega)^2}
{1+\varepsilon^2(4\omega^3-3\omega)^2}
}
$$

Esta expresión permite determinar la magnitud de la reflexión en función de la frecuencia.

### 2.4. Cálculo de los polos del filtro

Los polos de un filtro Chebyshev de tipo I se calculan mediante:

$$
p_k=-\sinh(\beta)\sin(\theta_k)
+j\cosh(\beta)\cos(\theta_k)
$$

donde:

$$
\beta=\frac{1}{n}\operatorname{arsinh}
\left(\frac{1}{\varepsilon}\right)
$$

y

$$
\theta_k=\frac{(2k-1)\pi}{2n}
$$

Para $n=3$ y $\varepsilon=0.50885$:

$$
\beta=\frac{1}{3}\operatorname{arsinh}
\left(\frac{1}{0.50885}\right)
$$

$$
\boxed{\beta\approx0.47599}
$$

Los ángulos correspondientes son:

$$
\theta_1=\frac{\pi}{6}
$$

$$
\theta_2=\frac{\pi}{2}
$$

$$
\theta_3=\frac{5\pi}{6}
$$

Reemplazando en la expresión de los polos, se obtiene aproximadamente:

$$
\boxed{p_1=-0.2471+j0.9660}
$$

$$
\boxed{p_2=-0.4942}
$$

$$
\boxed{p_3=-0.2471-j0.9660}
$$

Los tres polos se encuentran en el semiplano izquierdo del plano complejo, por lo que el filtro es estable.



### Obtención de $S_{11}(s)$

Para obtener la función de reflexión del cuadripolo, partimos de la expresión calculada anteriormente:

$$
|S_{11}(j\omega)|^2=
\frac{\varepsilon^2(4\omega^3-3\omega)^2}
{1+\varepsilon^2(4\omega^3-3\omega)^2}
$$

Como la red es ideal y sin pérdidas, se cumple:

$$
|S_{11}(j\omega)|^2+|S_{21}(j\omega)|^2=1
$$

Los ceros de reflexión se obtienen igualando a cero el polinomio de Chebyshev:

$$
T_3(\omega)=4\omega^3-3\omega=0
$$

Factorizando:

$$
\omega(4\omega^2-3)=0
$$

De donde:

$$
\omega_1=0
$$

$$
\omega_{2,3}=\pm\frac{\sqrt{3}}{2}
$$

Para pasar al dominio de Laplace, utilizamos:

$$
s=j\omega
$$

Por lo tanto, los ceros de $S_{11}(s)$ son:

$$
s_1=0
$$

$$
s_{2,3}=\pm j\frac{\sqrt{3}}{2}
$$

El polinomio del numerador resulta:

$$
N(s)=s\left(s-j\frac{\sqrt{3}}{2}\right)
\left(s+j\frac{\sqrt{3}}{2}\right)
$$

Aplicando diferencia de cuadrados:

$$
N(s)=s\left(s^2+\frac{3}{4}\right)
$$

Finalmente:

$$
\boxed{N(s)=s^3+0.75s}
$$

### 2.6. Construcción del denominador

A partir de los polos obtenidos anteriormente:

$$
p_1=-0.247085+j0.965999
$$

$$
p_2=-0.494171
$$

$$
p_3=-0.247085-j0.965999
$$

El denominador se construye mediante:

$$
D(s)=(s-p_1)(s-p_2)(s-p_3)
$$

Agrupando los polos complejos conjugados:

$$
D(s)=(s+0.494171)
\left[(s+0.247085)^2+0.965999^2\right]
$$

Desarrollando:

$$
\boxed{
D(s)=s^3+0.988341s^2+1.238409s+0.491307
}
$$

Por lo tanto, la función de reflexión del cuadripolo queda:

$$
\boxed{
S_{11}(s)=
\frac{s^3+0.75s}
{s^3+0.988341s^2+1.238409s+0.491307}
}
$$

Se adopta la solución de fase correspondiente a la red pasiva LC que se sintetizará posteriormente.

### 2.7. Obtención de la impedancia de entrada $Z_1(s)$

La relación entre el coeficiente de reflexión y la impedancia de entrada es:

$$
S_{11}(s)=\frac{Z_1(s)-Z_0}{Z_1(s)+Z_0}
$$

Despejando la impedancia:

$$
S_{11}(s)\left[Z_1(s)+Z_0\right]=Z_1(s)-Z_0
$$

$$
S_{11}(s)Z_1(s)+S_{11}(s)Z_0=Z_1(s)-Z_0
$$

Agrupando los términos que contienen $Z_1(s)$:

$$
Z_1(s)\left[1-S_{11}(s)\right]
=Z_0\left[1+S_{11}(s)\right]
$$

Por lo tanto:

$$
\boxed{
Z_1(s)=Z_0\frac{1+S_{11}(s)}{1-S_{11}(s)}
}
$$

Como la impedancia de referencia está normalizada a $Z_0=1\,\Omega$:

$$
Z_1(s)=\frac{1+S_{11}(s)}{1-S_{11}(s)}
$$

Sustituyendo la función de reflexión:

$$
Z_1(s)=
\frac{
1+\frac{s^3+0.75s}{D(s)}
}{
1-\frac{s^3+0.75s}{D(s)}
}
$$

Multiplicando numerador y denominador por $D(s)$:

$$
Z_1(s)=\frac{D(s)+N(s)}{D(s)-N(s)}
$$

Sumando los polinomios:

$$
D(s)+N(s)=
2s^3+0.988341s^2+1.988409s+0.491307
$$

Restando:

$$
D(s)-N(s)=
0.988341s^2+0.488409s+0.491307
$$

Finalmente, la impedancia de entrada normalizada es:

$$
\boxed{
Z_1(s)=
\frac{
2s^3+0.988341s^2+1.988409s+0.491307
}{
0.988341s^2+0.488409s+0.491307
}
}
$$

Esta función racional será utilizada en el siguiente punto para sintetizar el circuito pasivo mediante divisiones polinómicas sucesivas.



## Punto 2 — Síntesis del cuadripolo mediante el método de Cauer

### 2.8. Síntesis de la impedancia de entrada

A partir de la impedancia obtenida en el punto anterior:

$$
Z_1(s)=
\frac{
2s^3+0.988341s^2+1.988409s+0.491307
}{
0.988341s^2+0.488409s+0.491307
}
$$

Se realiza la síntesis mediante el **método de Cauer**, que consiste en desarrollar la impedancia como una fracción continua.

Este procedimiento permite identificar sucesivamente los elementos de una red escalera.

### 2.9. Primera división polinómica — Obtención de $L_1$

Dividimos el numerador de $Z_1(s)$ por su denominador:

$$
\frac{
2s^3+0.988341s^2+1.988409s+0.491307
}{
0.988341s^2+0.488409s+0.491307
}
$$

El primer término del cociente se obtiene dividiendo los términos de mayor grado:

$$
\frac{2s^3}{0.988341s^2}=2.02359s
$$

Por lo tanto:

$$
Z_1(s)=2.02359s+
\frac{
0.994204s+0.491307
}{
0.988341s^2+0.488409s+0.491307
}
$$

Como la impedancia de un inductor está dada por:

$$
Z_L(s)=sL
$$

identificamos:

$$
\boxed{L_1=2.02359\,\mathrm{H}}
$$

El primer elemento del circuito es un **inductor conectado en serie**.

### 2.10. Segunda división polinómica — Obtención de $C_2$

Luego de extraer el primer inductor, la impedancia restante es:

$$
Z_R(s)=
\frac{
0.994204s+0.491307
}{
0.988341s^2+0.488409s+0.491307
}
$$

Como el siguiente elemento corresponde a una rama en paralelo, trabajamos con la admitancia:

$$
Y_R(s)=\frac{1}{Z_R(s)}
$$

Por lo tanto:

$$
Y_R(s)=
\frac{
0.988341s^2+0.488409s+0.491307
}{
0.994204s+0.491307
}
$$

Realizando la división polinómica, se obtiene aproximadamente:

$$
Y_R(s)=0.99410s+
\frac{0.491307}{0.994204s+0.491307}
$$

La admitancia de un capacitor está dada por:

$$
Y_C(s)=sC
$$

Por lo tanto:

$$
\boxed{C_2=1.005\,\mathrm{F}}
$$

El segundo elemento del circuito es un **capacitor conectado en paralelo**.

### 2.11. Tercera división — Obtención de $L_3$

Después de extraer el capacitor, la admitancia restante es:

$$
Y_3(s)=
\frac{0.491307}{0.994204s+0.491307}
$$

Invertimos para obtener la impedancia:

$$
Z_3(s)=\frac{1}{Y_3(s)}
$$

$$
Z_3(s)=
\frac{0.994204s+0.491307}{0.491307}
$$

Separando términos:

$$
Z_3(s)\approx2.02359s+1
$$

Comparando con la impedancia de un inductor en serie con una resistencia:

$$
Z_3(s)=sL_3+R_L
$$

Obtenemos:

$$
\boxed{L_3=2.02359\,\mathrm{H}}
$$

$$
\boxed{R_L=1\,\Omega}
$$

### 2.12. Resultados de la síntesis

Los elementos normalizados del filtro son:

| Elemento | Valor normalizado | Conexión |
|---|---:|---|
| $L_1$ | $2.02359\,\mathrm{H}$ | Serie |
| $C_2$ | $1.005\,\mathrm{F}$ | Paralelo |
| $L_3$ | $2.02359\,\mathrm{H}$ | Serie |
| $R_L$ | $1\,\Omega$ | Carga |

La impedancia de entrada queda expresada como:

$$
\boxed{
Z_1(s)=sL_1+
\frac{1}{
sC_2+\frac{1}{sL_3+R_L}
}
}
$$

La red obtenida es un filtro pasabajos de tercer orden con estructura **L–C–L**.

Como $L_1=L_3$, el cuadripolo presenta una estructura simétrica, considerando las mismas impedancias de referencia en ambos puertos.



## Punto 3 — Simulación del filtro en LTspice

### 3.1. Implementación del circuito

Una vez sintetizado el filtro pasabajos Chebyshev de tercer orden, se procede a verificar su comportamiento mediante una simulación en frecuencia utilizando **LTspice**.

El circuito está constituido por una red escalera L–C–L, alimentada mediante una fuente de tensión alterna con una resistencia interna de $1\,\Omega$ y terminada en una resistencia de carga de $1\,\Omega$.

Los valores normalizados utilizados son:

| Componente | Valor |
|---|---:|
| $L_1$ | $2.013\,\mathrm{H}$ |
| $C_2$ | $1.005\,\mathrm{F}$ |
| $L_3$ | $2.013\,\mathrm{H}$ |
| $R_G$ | $1\,\Omega$ |
| $R_L$ | $1\,\Omega$ |

Se configura la fuente con una amplitud de pequeña señal:

$$
V_G=1\,\mathrm{V}
$$

Para analizar el comportamiento del filtro se realiza un barrido de frecuencia mediante el comando `.ac` de LTspice.

### 3.2. Obtención del parámetro $S_{11}$

El parámetro $S_{11}$ representa el coeficiente de reflexión en el puerto de entrada, cuando el puerto de salida se encuentra adaptado.

Se define como:

$$
S_{11}=\frac{Z_{\mathrm{in}}-Z_0}{Z_{\mathrm{in}}+Z_0}
$$

Considerando una fuente de tensión $V_G$ con resistencia interna $R_G=Z_0$, la tensión en el puerto de entrada es:

$$
V_1=V_G\frac{Z_{\mathrm{in}}}{Z_0+Z_{\mathrm{in}}}
$$

Despejando la relación con el coeficiente de reflexión:

$$
S_{11}=2\frac{V_1}{V_G}-1
$$

Como la fuente se configuró con $V_G=1\,\mathrm{V}$:

$$
\boxed{S_{11}=2V_1-1}
$$

Por lo tanto, en LTspice se utiliza la siguiente expresión:

`2*V(v1)-1`

Para visualizar su magnitud en decibeles:

$$
\boxed{S_{11,\mathrm{dB}}=20\log_{10}|S_{11}|}
$$

### 3.3. Obtención del parámetro $S_{21}$

El parámetro $S_{21}$ representa la transmisión desde el puerto de entrada hacia el puerto de salida.

Para una red con impedancias de referencia iguales y reales, con el puerto de salida adaptado:

$$
S_{21}=2\frac{V_2}{V_G}
$$

Como la fuente tiene una amplitud de $1\,\mathrm{V}$:

$$
\boxed{S_{21}=2V_2}
$$

En LTspice se utiliza la expresión:

`2*V(v2)`

Su magnitud en decibeles se calcula como:

$$
\boxed{S_{21,\mathrm{dB}}=20\log_{10}|S_{21}|}
$$

### 3.4. Interpretación de las respuestas

A partir de las curvas obtenidas se analiza:

- **Banda de paso:** $S_{21}$ presenta una transmisión elevada, con el rizado característico del filtro Chebyshev.
- **Frecuencia de corte:** se alcanza el borde de la banda de paso, donde la atenuación nominal es de $1\,\mathrm{dB}$.
- **Banda de transición:** disminuye la transmisión y aumenta la reflexión.
- **Banda detenida:** $S_{21}$ tiende a valores muy pequeños y $S_{11}$ se aproxima a $0\,\mathrm{dB}$.

Para una red ideal sin pérdidas se verifica:

$$
\boxed{|S_{11}|^2+|S_{21}|^2=1}
$$

Esta igualdad expresa la conservación de potencia entre la onda reflejada y la transmitida.

## Punto 4 — Análisis del parámetro $S_{11}$

El parámetro $S_{11}$ representa el coeficiente de reflexión en el puerto de entrada del cuadripolo.

Su magnitud permite determinar qué proporción de la potencia incidente es reflejada hacia el generador.

La relación entre la potencia reflejada y la potencia incidente es:

$$
\frac{P_{\mathrm{ref}}}{P_{\mathrm{inc}}}=|S_{11}|^2
$$

Cuando $S_{11}$ se expresa en decibeles:

$$
S_{11,\mathrm{dB}}=20\log_{10}|S_{11}|
$$

Por lo tanto, la fracción de potencia reflejada puede calcularse mediante:

$$
\boxed{
\frac{P_{\mathrm{ref}}}{P_{\mathrm{inc}}}
=10^{S_{11,\mathrm{dB}}/10}
}
$$

A continuación, se analiza el comportamiento del filtro en cuatro regiones de frecuencia.

### 4.1. Banda de paso

En la banda de paso, el filtro permite la transmisión de la señal con una atenuación reducida.

Para analizar esta región, se seleccionó una frecuencia aproximada de:

$$
f=79.454\,\mathrm{mHz}
$$

La simulación en LTspice arrojó:

$$
S_{11}=-6.8735\,\mathrm{dB}
$$

La fracción de potencia reflejada resulta:

$$
\frac{P_{\mathrm{ref}}}{P_{\mathrm{inc}}}
=10^{-6.8735/10}
\approx0.205
$$

Por lo tanto, aproximadamente el **20.5 % de la potencia incidente es reflejada**.

El resto de la potencia se transmite hacia la carga, considerando una red ideal sin pérdidas.

### 4.2. Frecuencia de corte

Para el filtro normalizado, la frecuencia angular del borde de la banda de paso es:

$$
\omega_p=1\,\mathrm{rad/s}
$$

Expresada en hertz:

$$
f_p=\frac{\omega_p}{2\pi}

S_{11}=-6.9391\,\mathrm{dB}

=10^{-6.9391/10}

Por lo tanto, aproximadamente el **20.2 % de la potencia incidente se refleja**.

## Punto 5 — Desnormalización del filtro Chebyshev

### 5.1. Datos del problema

Una vez sintetizado y analizado el filtro normalizado, se procede a realizar su desnormalización en frecuencia e impedancia.

Los valores normalizados obtenidos son:

| Componente | Valor normalizado |
|---|---:|
| $L_1$ | $2.013\,\mathrm{H}$ |
| $C_2$ | $1.005\,\mathrm{F}$ |
| $L_3$ | $2.013\,\mathrm{H}$ |
| $R_G$ | $1\,\Omega$ |
| $R_L$ | $1\,\Omega$ |

Se desea obtener un filtro con las siguientes características:

- Frecuencia de borde de la banda de paso: $f_p=1\,\mathrm{MHz}$
- Resistencia del generador: $R_G=50\,\Omega$
- Resistencia de carga: $R_L=50\,\Omega$

### 5.2. Factor de escalamiento de frecuencia

La frecuencia angular deseada es:

$$
\omega_p=2\pi f_p
$$

Reemplazando:

$$
\omega_p=2\pi\cdot10^6
$$

Por lo tanto:

$$
\boxed{\omega_p=6.283185\cdot10^6\,\mathrm{rad/s}}
$$

Como el filtro original está normalizado a $1\,\mathrm{rad/s}$, el factor de escalamiento de frecuencia resulta:

$$
\boxed{K_f=2\pi\cdot10^6}
$$

### 5.3. Factor de escalamiento de impedancia

La impedancia de referencia normalizada es:

$$
Z_0=1\,\Omega
$$

La nueva impedancia de referencia es:

$$
Z_0'=50\,\Omega
$$

Por lo tanto, el factor de escalamiento de impedancia es:

$$
K_Z=\frac{Z_0'}{Z_0}
$$

$$
\boxed{K_Z=50}
$$

### 5.4. Desnormalización de los inductores

La expresión para desnormalizar una inductancia es:

$$
\boxed{L'=\frac{K_Z}{K_f}L}
$$

Para el primer inductor:

$$
L_1'=\frac{50}{2\pi\cdot10^6}\cdot2.013
$$

$$
L_1'=1.6019\cdot10^{-5}\,\mathrm{H}
$$

Por lo tanto:

$$
\boxed{L_1'\approx16.02\,\mu\mathrm{H}}
$$

Como el filtro es simétrico y $L_1=L_3$:

$$
\boxed{L_3'\approx16.02\,\mu\mathrm{H}}
$$

### 5.5. Desnormalización del capacitor

La expresión para desnormalizar una capacitancia es:

$$
\boxed{C'=\frac{C}{K_ZK_f}}
$$

Reemplazando:

$$
C_2'=\frac{1.005}{50\cdot2\pi\cdot10^6}
$$

$$
C_2'=3.199\cdot10^{-9}\,\mathrm{F}
$$

Por lo tanto:

$$
\boxed{C_2'\approx3.199\,\mathrm{nF}}
$$

### 5.6. Desnormalización de las resistencias

Las resistencias se escalan mediante:

$$
R'=K_ZR
$$

Por lo tanto:

$$
R_G'=50\cdot1
$$

$$
R_L'=50\cdot1
$$

Finalmente:

$$
\boxed{R_G'=R_L'=50\,\Omega}
$$

### 5.7. Resultados del filtro desnormalizado

Los componentes del circuito resultante son:

| Componente | Valor desnormalizado |
|---|---:|
| $L_1$ | $16.02\,\mu\mathrm{H}$ |
| $C_2$ | $3.199\,\mathrm{nF}$ |
| $L_3$ | $16.02\,\mu\mathrm{H}$ |
| $R_G$ | $50\,\Omega$ |
| $R_L$ | $50\,\Omega$ |

El circuito conserva la estructura escalera L–C–L y presenta una frecuencia de borde de banda próxima a $1\,\mathrm{MHz}$.

### 5.8. Verificación mediante LTspice

Para verificar la desnormalización, se implementó nuevamente el circuito en LTspice, utilizando los valores calculados.

![Circuito desnormalizado](images/desnormalizado.png)

**Figura 8.** Implementación del filtro Chebyshev desnormalizado.

Se realiza un análisis de pequeña señal en frecuencia mediante:

`.ac dec 100 10k 10Meg`

Para observar la transmisión se utiliza:

`2*V(v2)`

La respuesta debe presentar una atenuación próxima a $1\,\mathrm{dB}$ en la frecuencia de borde de $1\,\mathrm{MHz}$.

Las pequeñas diferencias pueden deberse al redondeo de los componentes utilizados.

### 5.9. Conclusión

Mediante el proceso de desnormalización se obtuvo un filtro pasabajos Chebyshev de tercer orden, adaptado a impedancias de referencia de $50\,\Omega$ y con una frecuencia de borde de banda de $1\,\mathrm{MHz}$.

El procedimiento permite trasladar el comportamiento de un prototipo normalizado a las condiciones de frecuencia e impedancia requeridas, conservando la forma de su respuesta en frecuencia.



# Parte B — Investigación sobre NanoVNA

## 1. Función principal de un analizador vectorial de redes (VNA)

Un **analizador vectorial de redes (VNA)** es un instrumento que permite caracterizar el comportamiento de dispositivos y circuitos electrónicos en función de la frecuencia.

Su funcionamiento consiste en aplicar una señal de prueba al dispositivo bajo ensayo (DUT) y analizar las señales reflejadas y transmitidas, considerando tanto su **magnitud como su fase**.

El NanoVNA V2 permite medir principalmente dos parámetros de dispersión:

- **$S_{11}$:** coeficiente de reflexión en el puerto de entrada.
- **$S_{21}$:** coeficiente de transmisión desde el puerto de entrada hacia el puerto de salida.

A partir de estas mediciones es posible determinar:

- Impedancia compleja.
- Resistencia y reactancia.
- Pérdida de retorno.
- Pérdida de inserción.
- Relación de ondas estacionarias (ROE o VSWR).

### 1.1. Aplicaciones principales

Un VNA resulta especialmente útil en el análisis de circuitos de radiofrecuencia (RF), donde es importante conocer la adaptación de impedancias y la transmisión de señales.

Entre sus aplicaciones se encuentran:

- **Antenas:** determinar su adaptación de impedancia y frecuencia de resonancia.
- **Filtros:** medir la respuesta en frecuencia, las pérdidas de inserción y la reflexión.
- **Amplificadores:** caracterizar su respuesta de transmisión y adaptación.
- **Cables y líneas de transmisión:** detectar discontinuidades y evaluar su comportamiento.
- **Duplexores:** analizar la transmisión y el aislamiento entre sus caminos de señal.

### 1.2. Importancia del VNA en el diseño de filtros

En el caso del filtro Chebyshev desarrollado en la Parte A, un VNA permitiría medir experimentalmente los parámetros $S_{11}$ y $S_{21}$.

Esto permitiría comparar las mediciones reales con las respuestas obtenidas mediante LTspice, verificando:

- La frecuencia de borde de la banda de paso.
- El rizado y la atenuación de transmisión.
- La adaptación de impedancias.
- El comportamiento del filtro fuera de la banda de paso.

Por lo tanto, el VNA constituye una herramienta fundamental para verificar el funcionamiento de filtros y otros circuitos de RF.

## 2. Comparación entre NanoVNA V2 y osciloscopio GW Instek

El NanoVNA V2 y el osciloscopio digital GW Instek GDS-1000A-U son instrumentos de medición electrónica, pero presentan diferencias importantes en su funcionamiento y aplicaciones.

El **NanoVNA** está diseñado para caracterizar redes y dispositivos mediante parámetros de dispersión en función de la frecuencia.

En cambio, el **osciloscopio** permite observar señales eléctricas en el dominio del tiempo, midiendo su amplitud, período, frecuencia y otras características.

### 2.1. Comparación de características técnicas

| Característica | NanoVNA V2 | GW Instek GDS-1000A-U |
|---|---|---|
| Función principal | Análisis de redes de RF | Visualización de señales eléctricas |
| Rango de frecuencias | 50 kHz a 3 GHz o 4.4 GHz, según versión | DC a 70, 100 o 150 MHz, según modelo |
| Rango dinámico | 70 dB a 1 GHz (V2 original); hasta 90 dB a 1 GHz (Plus4) | No especificado directamente en dB |
| Impedancia de referencia/entrada | 50 $\Omega$ nominales en los puertos RF | 1 M$\Omega$ $\pm 2\%$, aproximadamente 15 pF |
| Magnitudes medidas | Parámetros $S_{11}$ y $S_{21}$, magnitud y fase | Tensión en función del tiempo |
| Resolución vertical | No corresponde a una resolución de osciloscopio | 8 bits |
| Muestreo en tiempo real | No se especifica como un osciloscopio | Hasta 1 GSa/s |

### 2.2. Ancho de banda

El NanoVNA V2 permite realizar barridos de frecuencia en un rango considerablemente más amplio que el ancho de banda de los osciloscopios de la serie GDS-1000A-U.

Sin embargo, estas especificaciones no son directamente equivalentes:

- En el NanoVNA se indica el intervalo de frecuencias en el cual puede caracterizar dispositivos.
- En el osciloscopio se indica el ancho de banda de su sistema de adquisición, habitualmente definido por una caída de 3 dB.

### 2.3. Rango dinámico

El rango dinámico representa la diferencia entre los niveles de señal más altos y más bajos que pueden medirse adecuadamente.

En el NanoVNA V2, este parámetro es especialmente importante para medir atenuaciones elevadas, por ejemplo, en la banda detenida de un filtro.

La versión V2 original presenta un rango dinámico calibrado de aproximadamente 70 dB a 1 GHz, mientras que la versión Plus4 alcanza 90 dB bajo determinadas condiciones de medición.

El manual del GW Instek GDS-1000A-U no proporciona directamente una especificación de rango dinámico en dB.

Aunque el osciloscopio utiliza una resolución vertical de 8 bits, este dato no debe interpretarse automáticamente como su rango dinámico efectivo.

### 2.4. Impedancia de entrada

El NanoVNA utiliza puertos de RF con una impedancia de referencia nominal de:

$$
\boxed{Z_0=50\,\Omega}
$$

Esta característica facilita las mediciones de adaptación de impedancias en circuitos de radiofrecuencia.

Por otro lado, el osciloscopio GW Instek presenta una impedancia de entrada de:

$$
\boxed{Z_{\mathrm{in}}\approx1\,\mathrm{M}\Omega\parallel15\,\mathrm{pF}}
$$

La elevada resistencia de entrada permite medir tensiones reduciendo la carga resistiva sobre el circuito, aunque la capacitancia de entrada también debe considerarse, especialmente a frecuencias elevadas.

### 2.5. Conclusión de la comparación

Ambos instrumentos cumplen funciones diferentes y complementarias.

El NanoVNA resulta más adecuado para medir la respuesta en frecuencia, la transmisión y la reflexión de dispositivos de RF, como filtros, antenas y líneas de transmisión.

El osciloscopio GW Instek resulta más adecuado para visualizar señales en el tiempo y analizar fenómenos como transitorios, pulsos y variaciones de tensión.

Para caracterizar experimentalmente el filtro Chebyshev desarrollado en la Parte A, el NanoVNA permitiría medir directamente los parámetros $S_{11}$ y $S_{21}$ dentro de su rango de funcionamiento.

## 3. Mediciones en el dominio del tiempo con NanoVNA V2

### 3.1. ¿El NanoVNA permite realizar mediciones en el dominio del tiempo?

**Sí.** Además de medir parámetros de dispersión en el dominio de la frecuencia, el NanoVNA V2 permite representar información en el dominio del tiempo.

Según su manual de usuario, el instrumento puede simular mediciones de reflectometría en el dominio del tiempo, conocidas como **TDR (Time Domain Reflectometry)**.

Para ello, transforma matemáticamente los datos obtenidos durante el barrido de frecuencias.

Es importante destacar que el NanoVNA no mide directamente una señal temporal como lo hace un osciloscopio, sino que obtiene una representación temporal a partir de sus mediciones en frecuencia.

### 3.2. Principio de funcionamiento

El NanoVNA mide los parámetros de dispersión en función de la frecuencia.

Posteriormente, mediante una transformación matemática relacionada con la transformada inversa de Fourier, es posible obtener una representación de la respuesta temporal del dispositivo bajo ensayo.

De manera conceptual:

$$
\boxed{S_{11}(f)\longrightarrow s_{11}(t)}
$$

La respuesta temporal permite identificar reflexiones asociadas a discontinuidades de impedancia.

Por ejemplo, en un cable coaxial pueden aparecer reflexiones debido a:

- Conectores defectuosos.
- Cambios de impedancia.
- Circuitos abiertos.
- Cortocircuitos.
- Discontinuidades capacitivas o inductivas.

### 3.3. Activación de la medición temporal

Según el manual del NanoVNA V2, la transformación al dominio del tiempo se habilita mediante la siguiente secuencia:

`DISPLAY → TRANSFORM → TRANSFORM ON`

Una vez activada, el equipo convierte los datos medidos en frecuencia y presenta su representación temporal.

### 3.4. Modos de medición disponibles

El manual describe tres modos principales:

**a) Bandpass**

Permite representar una respuesta impulsional del dispositivo bajo ensayo.

Puede utilizarse, por ejemplo, para analizar la respuesta temporal de un filtro pasabanda.

**b) Low Pass Impulse**

Permite obtener una representación de la respuesta al impulso.

En este modo puede utilizarse la medición de reflexión para realizar análisis de tipo TDR.

**c) Low Pass Step**

Permite obtener una representación de la respuesta al escalón.

Es útil para analizar discontinuidades de impedancia, como circuitos abiertos, cortocircuitos y comportamientos capacitivos o inductivos.

### 3.5. Relación entre tiempo y distancia

Una de las aplicaciones de la reflectometría temporal es estimar la ubicación de discontinuidades en cables de transmisión.

La distancia puede calcularse mediante:

$$
\boxed{d=\frac{v_p t}{2}}
$$

donde:

- $d$ es la distancia hasta la discontinuidad.
- $v_p$ es la velocidad de propagación de la señal en el cable.
- $t$ es el tiempo de ida y vuelta de la reflexión.

La velocidad de propagación puede expresarse como:

$$
v_p=VF\cdot c
$$

donde:

- $VF$ es el factor de velocidad del cable.
- $c$ es la velocidad de la luz en el vacío.

Por lo tanto:

$$
\boxed{d=\frac{VF\cdot c\cdot t}{2}}
$$

El NanoVNA permite configurar el factor de velocidad para representar la información temporal en términos de distancia.

### 3.6. Limitaciones de la medición temporal

La precisión de la representación temporal depende de las características del barrido de frecuencia.

Según el manual:

- Un mayor intervalo de frecuencias mejora la resolución temporal.
- Un menor espaciamiento entre los puntos de frecuencia permite aumentar el intervalo máximo de tiempo observable.
- Existe un compromiso entre resolución temporal y alcance de medición.

Además, el equipo permite utilizar ventanas de procesamiento para reducir las oscilaciones artificiales producidas por el rango finito de frecuencias.

### 3.7. Conclusión

El NanoVNA V2 permite realizar análisis en el dominio del tiempo mediante la transformación matemática de sus mediciones en frecuencia.

Esta funcionalidad es especialmente útil para analizar líneas de transmisión, identificar discontinuidades de impedancia y estimar su ubicación.

Por lo tanto, además de caracterizar filtros y circuitos mediante los parámetros $S_{11}$ y $S_{21}$, el NanoVNA puede utilizarse como una herramienta de reflectometría temporal.
