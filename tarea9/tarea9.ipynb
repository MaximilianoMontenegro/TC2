# Síntesis de cuadripolos pasivos LC

##  Introducción teórica

La **síntesis de redes eléctricas** consiste en determinar un circuito que reproduzca una función de transferencia previamente especificada.

A diferencia del análisis de circuitos, donde se parte de los componentes para obtener su comportamiento, en la síntesis se parte de una expresión matemática para determinar una posible realización física.

En este trabajo se estudia la síntesis de **cuadripolos pasivos LC**, constituidos por inductores y capacitores ideales, mediante técnicas de remoción de polos y análisis de funciones de transferencia.

###  Función de transferencia

La función de transferencia relaciona una variable de salida con una variable de entrada en el dominio de Laplace, suponiendo condiciones iniciales nulas.

En los ejercicios considerados se trabaja con una **transferencia de transimpedancia**:

$$
T(s)=\frac{V_2(s)}{I_g(s)}
$$

donde:

- $V_2(s)$ es la tensión de salida.
- $I_g(s)$ es la corriente del generador.
- $s=\sigma+j\omega$ es la variable compleja de Laplace.

###  Parámetros de cuadripolos

Un cuadripolo puede caracterizarse mediante distintos conjuntos de parámetros, entre ellos los parámetros de impedancia $Z$ y los parámetros de transmisión $ABCD$.

**Parámetros de impedancia:**

$$
\begin{bmatrix}
V_1\\
V_2
\end{bmatrix}
=
\begin{bmatrix}
Z_{11} & Z_{12}\\
Z_{21} & Z_{22}
\end{bmatrix}
\begin{bmatrix}
I_1\\
I_2
\end{bmatrix}
$$

**Parámetros de transmisión:**

$$
\begin{bmatrix}
V_1\\
I_1
\end{bmatrix}
=
\begin{bmatrix}
A & B\\
C & D
\end{bmatrix}
\begin{bmatrix}
V_2\\
-I_2
\end{bmatrix}
$$

Los parámetros de transmisión son especialmente útiles para analizar circuitos formados por etapas conectadas en cascada.

###  Método de síntesis

Para obtener una realización LC se emplean técnicas de **remoción de polos**, mediante las cuales se extraen sucesivamente elementos reactivos de una función de impedancia o admitancia.

Las impedancias elementales son:

**Inductor:**

$$
Z_L(s)=sL
$$

**Capacitor:**

$$
Z_C(s)=\frac{1}{sC}
$$

Las admitancias correspondientes son:

$$
Y_L(s)=\frac{1}{sL}
$$

$$
Y_C(s)=sC
$$

Las remociones permiten identificar elementos en serie o en derivación, según se trabaje con impedancias o admitancias.

###  Verificación de la realización

Una vez sintetizado el circuito, se calcula nuevamente su función de transferencia y se compara con la expresión solicitada.

Para estudiar su respuesta en frecuencia se realiza la sustitución:

$$
s=j\omega
$$

Esto permite identificar:

- Polos y ceros de transmisión.
- Frecuencias de resonancia.
- Bandas de paso y de rechazo.
- Frecuencias de corte.
- Ganancia expresada en decibeles.

La magnitud de una transferencia normalizada puede expresarse en decibeles como:

$$
G_{dB}(\omega)=20\log_{10}|T(j\omega)|
$$

Cuando $T(s)$ representa una transimpedancia, debe especificarse una impedancia de referencia para expresar su magnitud en dB.



## 2. Ejercicio 1 — Síntesis mediante parámetros Z
## A) Sintetitzar cuadripolo

### Función de transferencia

Se considera un cuadripolo LC alimentado por un generador de corriente $I_g$, con una resistencia $R_g$ en paralelo y la salida en circuito abierto.

La transferencia solicitada es:

$$
T(s)=\frac{V_2}{I_g}
=\frac{k\,s(s^2+1/16)}{s^3+2s^2+2s+1}
$$

Utilizando parámetros de impedancia y considerando $I_2=0$:

$$
V_1=Z_{11}I_1,\qquad V_2=Z_{21}I_1
$$

La corriente de entrada resulta:

$$
I_1=I_g\frac{R_g}{R_g+Z_{11}}
$$

Por lo tanto:

$$
\boxed{
T(s)=\frac{Z_{21}}{1+Z_{11}/R_g}
}
$$

Normalizando $R_g=1$:

$$
T(s)=\frac{Z_{21}}{1+Z_{11}}
$$

###  Obtención de las funciones Z

Separando el denominador en sus partes par e impar:

$$
D(s)=(2s^2+1)+(s^3+2s)
$$

Se proponen:

$$
\boxed{
Z_{11}(s)=\frac{s(s^2+2)}{2s^2+1}
}
$$

$$
\boxed{
Z_{21}(s)=\frac{k\,s(s^2+1/16)}{2s^2+1}
}
$$

Estas expresiones permiten recuperar la transferencia original.

###  Primera remoción: inductor en derivación

Se invierte la impedancia de entrada:

$$
Y_{11}(s)=\frac{1}{Z_{11}(s)}
=\frac{2s^2+1}{s(s^2+2)}
$$

Se realiza una remoción parcial en el origen para generar un cero en $s=\pm j/4$:

$$
k_1=
\left.
\frac{2s^2+1}{s^2+2}
\right|_{s^2=-1/16}
=\frac{14}{31}
$$

La admitancia removida es:

$$
Y_L(s)=\frac{14}{31s}
$$

Por lo tanto, el inductor es:

$$
\boxed{L_1=\frac{31}{14}}
$$

La admitancia residual resulta:

$$
Y_1(s)=Y_{11}(s)-\frac{14}{31s}
$$

$$
\boxed{
Y_1(s)=\frac{48}{31}
\frac{s^2+1/16}{s(s^2+2)}
}
$$

###  Segunda remoción: tanque LC paralelo

Se invierte la admitancia residual:

$$
Z_1(s)=\frac{1}{Y_1(s)}
$$

$$
\boxed{
Z_1(s)=\frac{31}{48}
\frac{s(s^2+2)}{s^2+1/16}
}
$$

Esta impedancia presenta polos en:

$$
s=\pm j\frac14
$$

Se remueve el término correspondiente al tanque LC:

$$
2k'_1=
\left.
\frac{s^2+1/16}{s}Z_1(s)
\right|_{s^2=-1/16}
$$

$$
2k'_1=\frac{31}{48}\left(2-\frac1{16}\right)
=\frac{961}{768}
$$

La impedancia del tanque es:

$$
Z_T(s)=\frac{961}{768}\frac{s}{s^2+1/16}
$$

Comparando con la impedancia de un tanque LC paralelo:

$$
Z_T(s)=\frac{1}{C_T}\frac{s}{s^2+1/(L_TC_T)}
$$

Se obtienen:

$$
\boxed{C_T=\frac{768}{961}}
$$

$$
\boxed{L_T=\frac{961}{48}}
$$

###  Tercera remoción: inductor residual

Se resta la impedancia del tanque:

$$
Z_2(s)=Z_1(s)-Z_T(s)
$$

Simplificando:

$$
\boxed{Z_2(s)=\frac{31}{48}s}
$$

Por lo tanto, corresponde a un inductor:

$$
\boxed{L_2=\frac{31}{48}}
$$

###  Resultados de la síntesis

Los elementos obtenidos son:

| Elemento | Valor normalizado | Ubicación |
|---|---|---|
| $L_1$ | $31/14$ | Derivación de entrada |
| $L_T$ | $961/48$ | Tanque LC paralelo |
| $C_T$ | $768/961$ | Tanque LC paralelo |
| $L_2$ | $31/48$ | Derivación de salida |

La red resultante presenta una estructura en $\pi$, formada por dos inductores en derivación y un tanque LC paralelo conectado en serie entre ambos puertos.

La verificación de la transferencia permitirá comprobar la realización y determinar la constante $k$.


## B )Determinación de la constante de transferencia $k$

La función de transferencia del circuito es:

$$
T(s)=\frac{V_2(s)}{V_g(s)}
=k\frac{s(s^2+1/16)}{s^3+2s^2+2s+1}
$$

Para determinar la constante $k$, se analiza el comportamiento del circuito en altas frecuencias:

$$
\lim_{s\to\infty}T(s)=k
$$

Recordando las impedancias de los elementos reactivos:

$$
Z_L=sL,\qquad Z_C=\frac{1}{sC}
$$

Cuando la frecuencia tiende a infinito:

- Los inductores se comportan como circuitos abiertos.
- Los capacitores se comportan como cortocircuitos.

En consecuencia, el inductor de entrada y el de salida quedan abiertos, mientras que el capacitor del tanque conecta ambos puertos.

Como la salida está en circuito abierto, no circula corriente por la resistencia del generador y no existe caída de tensión sobre ella.

Por lo tanto:

$$
V_2=V_g
$$

De esta manera:

$$
\lim_{s\to\infty}\frac{V_2}{V_g}=1
$$

Comparando con el límite de la transferencia:

$$
\boxed{k=1}
$$

**Conclusión:** la constante de transferencia del circuito sintetizado es $k=1$, lo que corresponde a una ganancia asintótica de $0\,\mathrm{dB}$ respecto de $V_g$.

## Ejercicio 2 — Síntesis de un cuadripolo LC con carga resistiva

### A) Función de transferencia

Se considera un cuadripolo pasivo LC alimentado por una fuente ideal de corriente $I_1$, con una resistencia de carga $R_L$ conectada a la salida.

La transferencia solicitada es:

$$
T(s)=\frac{V_2}{I_1}
=\frac{k(s^2+9)}{s^3+2s^2+2s+1}
$$

Utilizando parámetros de impedancia:

$$
V_2=Z_{21}I_1+Z_{22}I_2
$$

Como la salida está cargada:

$$
I_2=-\frac{V_2}{R_L}
$$

Por lo tanto:

$$
V_2\left(1+\frac{Z_{22}}{R_L}\right)=Z_{21}I_1
$$

Normalizando $R_L=1$:

$$
\boxed{T(s)=\frac{Z_{21}}{1+Z_{22}}}
$$

### Obtención de las funciones de impedancia

Separando el denominador en sus partes par e impar:

$$
D(s)=(2s^2+1)+s(s^2+2)
$$

Se proponen las siguientes funciones:

$$
\boxed{Z_{22}(s)=\frac{2s^2+1}{s(s^2+2)}}
$$

$$
\boxed{Z_{21}(s)=k\frac{s^2+9}{s(s^2+2)}}
$$

### Primera remoción: capacitor de salida

Se invierte la impedancia:

$$
Y_{22}(s)=\frac{1}{Z_{22}(s)}
=\frac{s(s^2+2)}{2s^2+1}
$$

Se realiza una remoción parcial en infinito, generando un cero en $s^2=-9$:

$$
C_1=
\left.
\frac{s^2+2}{2s^2+1}
\right|_{s^2=-9}
$$

$$
C_1=\frac{-9+2}{2(-9)+1}
=\frac{7}{17}
$$

Por lo tanto, el capacitor de salida es:

$$
\boxed{C_1=\frac{7}{17}}
$$

La admitancia residual resulta:

$$
Y_1(s)=Y_{22}(s)-\frac{7}{17}s
$$

Simplificando:

$$
\boxed{
Y_1(s)=\frac{3s(s^2+9)}{17(2s^2+1)}
}
$$

### Segunda remoción: tanque LC paralelo

Se invierte la admitancia residual:

$$
Z_1(s)=\frac{1}{Y_1(s)}
$$

$$
\boxed{
Z_1(s)=\frac{17(2s^2+1)}{3s(s^2+9)}
}
$$

Esta impedancia presenta polos en:

$$
s=\pm j3
$$

Se remueve el término correspondiente al tanque LC:

$$
2k_1=
\left.
\frac{s^2+9}{s}Z_1(s)
\right|_{s^2=-9}
$$

$$
2k_1=\frac{17}{3}
\left(\frac{2(-9)+1}{-9}\right)
$$

$$
\boxed{2k_1=\frac{289}{27}}
$$

La impedancia del tanque es:

$$
Z_T(s)=\frac{289}{27}\frac{s}{s^2+9}
$$

Comparando con la expresión de un tanque LC paralelo:

$$
Z_T(s)=\frac{1}{C_T}
\frac{s}{s^2+1/(L_TC_T)}
$$

Se obtiene:

$$
\boxed{C_T=\frac{27}{289}}
$$

$$
\boxed{L_T=\frac{289}{243}}
$$

### Tercera remoción: capacitor residual

Se resta la impedancia del tanque:

$$
Z_2(s)=Z_1(s)-Z_T(s)
$$

$$
Z_2(s)=
\frac{17(2s^2+1)}{3s(s^2+9)}
-\frac{289s}{27(s^2+9)}
$$

Simplificando:

$$
\boxed{Z_2(s)=\frac{17}{27s}}
$$

Esta expresión corresponde a la impedancia de un capacitor:

$$
Z_C(s)=\frac{1}{sC}
$$

Por lo tanto:

$$
\boxed{C_2=\frac{27}{17}}
$$

En la realización completa del cuadripolo, este capacitor se conecta en derivación en el puerto de entrada.

### Resultados de la síntesis

| Elemento | Valor normalizado | Ubicación |
|---|---|---|
| $C_2$ | $27/17$ | Derivación de entrada |
| $L_T$ | $289/243$ | Tanque LC paralelo |
| $C_T$ | $27/289$ | Tanque LC paralelo |
| $C_1$ | $7/17$ | Derivación de salida |
| $R_L$ | $1$ | Carga resistiva |

La red sintetizada presenta una estructura en $\pi$, formada por dos capacitores en derivación y un tanque LC paralelo conectado en serie entre ambos puertos.

La verificación de la transferencia permite determinar el valor de la constante $k$.



## B) Verificación de la transferencia mediante parámetros ABCD

Para verificar el circuito sintetizado, se utilizan los parámetros de transmisión:

$$
\begin{bmatrix}
V_1\\
I_1
\end{bmatrix}
=
\begin{bmatrix}
A&B\\
C&D
\end{bmatrix}
\begin{bmatrix}
V_2\\
-I_2
\end{bmatrix}
$$

Como la carga es $R_L=1$, se cumple:

$$
-I_2=\frac{V_2}{R_L}=V_2
$$

Por lo tanto:

$$
I_1=CV_2+DV_2
$$

La transferencia resulta:

$$
\boxed{
T(s)=\frac{V_2}{I_1}=\frac{1}{C+D}
}
$$

### Matriz de transmisión del circuito

El cuadripolo está formado por dos capacitores en derivación y un tanque LC paralelo conectado en serie.

La matriz total se obtiene multiplicando las matrices de cada etapa:

$$
\begin{bmatrix}
A&B\\
C&D
\end{bmatrix}
=
\begin{bmatrix}
1&0\\
Y_2&1
\end{bmatrix}
\begin{bmatrix}
1&Z_T\\
0&1
\end{bmatrix}
\begin{bmatrix}
1&0\\
Y_1&1
\end{bmatrix}
$$

De esta multiplicación se obtiene:

$$
C=Y_2+Y_1(1+Y_2Z_T)
$$

$$
D=1+Y_2Z_T
$$

Por lo tanto:

$$
\boxed{
C+D=1+Y_2Z_T+Y_2+Y_1(1+Y_2Z_T)
}
$$

### Sustitución de los componentes

Los valores obtenidos en la síntesis son:

$$
Y_1(s)=\frac{7}{17}s
$$

$$
Y_2(s)=\frac{27}{17}s
$$

$$
Z_T(s)=\frac{289}{27}\frac{s}{s^2+9}
$$

Sustituyendo:

$$
C+D=
1+\frac{17s^2}{s^2+9}
+\frac{27}{17}s
+\frac{7}{17}s
\left(1+\frac{17s^2}{s^2+9}\right)
$$

Agrupando y simplificando:

$$
C+D=
\frac{9s^3+18s^2+18s+9}{s^2+9}
$$

Por lo tanto:

$$
T(s)=\frac{1}{C+D}
$$

$$
T(s)=
\frac{s^2+9}{9s^3+18s^2+18s+9}
$$

Factorizando el denominador:

$$
\boxed{
T(s)=\frac19
\frac{s^2+9}{s^3+2s^2+2s+1}
}
$$

### Determinación de la constante k

Comparando con la transferencia solicitada:

$$
T(s)=k\frac{s^2+9}{s^3+2s^2+2s+1}
$$

Se obtiene:

$$
\boxed{k=\frac19}
$$

**Conclusión:** el circuito LC sintetizado reproduce exactamente la transferencia solicitada, con una constante de proporcionalidad $k=1/9$.

## Ejercicio 3 — Síntesis de un cuadripolo LC

Se considera la siguiente función de transferencia:

$$
T(s)=\frac{V_2(s)}{I_g(s)}
=\frac{k\,s\left(s^2+\frac{1}{16}\right)}
{s^3+2s^2+2s+1}
$$

El generador de corriente posee una resistencia interna $R_g$
en paralelo. Se busca sintetizar un cuadripolo pasivo sin pérdidas,
considerando la salida en circuito abierto.

### Planteo mediante parámetros de impedancia

Para un cuadripolo descrito mediante parámetros $Z$:

$$
V_1=Z_{11}I_1+Z_{12}I_2
$$

$$
V_2=Z_{21}I_1+Z_{22}I_2
$$

Como la salida está abierta:

$$
I_2=0
$$

Por división de corriente:

$$
I_1=I_g\frac{R_g}{R_g+Z_{11}}
$$

Por lo tanto:

$$
T(s)=\frac{V_2}{I_g}
=\frac{Z_{21}}{1+Z_{11}/R_g}
$$

Normalizando $R_g=1$:

$$
\boxed{T(s)=\frac{Z_{21}}{1+Z_{11}}}
$$

### Identificación de las funciones de impedancia

Separando los términos pares e impares del denominador:

$$
s^3+2s^2+2s+1=(2s^2+1)+(s^3+2s)
$$

Se proponen:

$$
\boxed{Z_{11}(s)=\frac{s(s^2+2)}{2s^2+1}}
$$

$$
\boxed{Z_{21}(s)=
\frac{k\,s\left(s^2+\frac{1}{16}\right)}{2s^2+1}}
$$

Estas expresiones permiten recuperar la transferencia original.

### Primera remoción: polo en el origen

Se invierte la impedancia de entrada:

$$
Y_{11}(s)=\frac{1}{Z_{11}(s)}
=\frac{2s^2+1}{s(s^2+2)}
$$

Se realiza una remoción parcial para generar un cero en:

$$
s^2=-\frac{1}{16}
$$

El coeficiente extraído es:

$$
k_1=
\left.
\frac{2s^2+1}{s^2+2}
\right|_{s^2=-1/16}
=\frac{14}{31}
$$

Por lo tanto:

$$
Y_1(s)=Y_{11}(s)-\frac{14}{31s}
$$

$$
\boxed{
Y_1(s)=\frac{48}{31}
\frac{s^2+\frac{1}{16}}{s(s^2+2)}
}
$$

El elemento extraído es un inductor en derivación:

$$
\boxed{L_1=\frac{31}{14}}
$$

### Segunda remoción: tanque LC paralelo

Se invierte la admitancia residual:

$$
Z_1(s)=\frac{1}{Y_1(s)}
=\frac{31}{48}
\frac{s(s^2+2)}{s^2+\frac{1}{16}}
$$

Esta función presenta polos en:

$$
s=\pm j\frac14
$$

Se extrae un tanque LC paralelo cuya impedancia es:

$$
Z_T(s)=2k_i\frac{s}{s^2+\frac{1}{16}}
$$

El coeficiente resulta:

$$
2k_i=
\frac{31}{48}
\left(2-\frac{1}{16}\right)
=\frac{961}{768}
$$

Comparando con la impedancia de un tanque paralelo:

$$
Z_T(s)=\frac{1}{C_T}
\frac{s}{s^2+\frac{1}{L_TC_T}}
$$

Se obtiene:

$$
\boxed{C_T=\frac{768}{961}}
$$

$$
\boxed{L_T=\frac{15376}{768}}
$$

### Tercera remoción: inductor residual

Se resta la impedancia del tanque:

$$
Z_2(s)=Z_1(s)-Z_T(s)
$$

$$
Z_2(s)=
\frac{31}{48}\frac{s(s^2+2)}{s^2+\frac{1}{16}}
-\frac{961}{768}\frac{s}{s^2+\frac{1}{16}}
$$

Simplificando:

$$
\boxed{Z_2(s)=\frac{31}{48}s}
$$

La impedancia residual corresponde a un inductor:

$$
\boxed{L_2=\frac{31}{48}}
$$

### Componentes obtenidos

| Elemento | Valor normalizado |
|---|---|
| Inductor en derivación $L_1$ | $31/14$ |
| Capacitor del tanque $C_T$ | $768/961$ |
| Inductor del tanque $L_T$ | $15376/768$ |
| Inductor residual $L_2$ | $31/48$ |

La red resultante está formada por un inductor en derivación,
un tanque LC paralelo y un inductor residual.



### Determinación de la constante $k$

Para evaluar la ganancia de la función de transferencia, se adopta
provisoriamente la frecuencia angular:

$$
\omega=1\ \text{rad/s}
$$

Se desea obtener una ganancia de $-6\,\mathrm{dB}$:

$$
20\log_{10}|T(j\omega)|=-6
$$

Por lo tanto:

$$
|T(j\omega)|=10^{-6/20}\approx 0.5012
$$

La función de transferencia es:

$$
T(s)=\frac{k\,s\left(s^2+\frac{1}{16}\right)}
{s^3+2s^2+2s+1}
$$

Evaluando en $s=j$:

$$
|T(j)|=
k\frac{15}{16\sqrt{2}}
$$

Despejando:

$$
\boxed{k=10^{-6/20}\frac{16\sqrt{2}}{15}}
$$

$$
\boxed{k\approx 0.756}
$$
